# Cyclic Peptide Modeller Demonstration

This notebook demonstrates the usage of the `CyclicPeptideModeller` to generate and optimise macrocyclic peptide conformations.

The underlying kinematic closure algorithm is described in [10.1002/jcc.10416](https://onlinelibrary.wiley.com/doi/abs/10.1002/jcc.10416).


```python
import numpy as np
import matplotlib.pyplot as plt
from loopmover.optimiser import CyclicPeptideOptimiser

# Example sequence (mix of L and D amino acids)
sequence = 'rQpqRePQ'
print(f"Target sequence: {sequence}")
```

## 1. Initialise and Optimise
We use the `CyclicPeptideOptimiser` to find low-energy states. Here we run a very short optimisation to demonstrate.


```python
# Instantiate optimiser
# Using a small number of iterations for demonstration purposes
optimiser = CyclicPeptideOptimiser(sequence)
optimiser.build_start_mac()
optimiser.amber_setup()
optimiser.optimise(n_iter=5, samplesize=20, max_iter=5, plot=False)

print(f"Found {len(optimiser.halloffame)} low energy states.")
```

## 2. Review Output
We can plot the energy of the sampled states over the optimisation trajectory.


```python
energies = [entry[0] for entry in optimiser.halloffame]
plt.figure(figsize=(6, 4))
plt.plot(energies, marker='o')
plt.title('Optimisation Energy')
plt.xlabel('State Index')
plt.ylabel('Energy (kJ/mol)')
plt.grid(True)
plt.show()
```
